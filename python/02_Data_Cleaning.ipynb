# 02 Data Cleaning

```python
import pandas as pd

df = pd.read_excel("../data/raw/Ecommerce_Retail_20K.xlsx")

print(df.shape)
df.head()
```

```python
# Missing values in the dataset
df.isnull().sum()
```

```python
# Duplicated rows in the dataset
df.duplicated().sum()
df=df.drop_duplicates()
```

```python
#Convert the InvoiceDate column to datetime format
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['CustomerID']=df['CustomerID'].astype('Int64')
```

```python
type(df['InvoiceDate'].iloc[0])
```

```python
# Feature Engineering 
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)
df['IsCancelled'] = df['InvoiceNo'].astype(str).str.startswith('C')
```

```python
# Build the clean dataframe (mirrors your SQL retail_clean view)
df_clean = df[
    (df['Quantity'] > 0) &
    (df['UnitPrice'] > 0) &
    (~df['IsCancelled']) &
    (~df['StockCode'].isin(['POST','DOT','M','BANK CHARGES','C2','PADS','CRUK']))
].copy()
```

```python
print(f"Raw rows: {len(df)}, Clean rows: {len(df_clean)}")
df_clean.to_csv("C:\\Users\\Siddhesh\\Downloads\\Projects\\E-commerce\\ecommerce-sales-customer-intelligence\\data\\retail_clean.csv", index=False)
```
