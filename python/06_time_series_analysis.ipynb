```python
import pandas as pd

df_clean = pd.read_csv("C:\\Users\\Siddhesh\\Downloads\\Projects\\E-commerce\\ecommerce-sales-customer-intelligence\\data\\retail_clean.csv")

print(df_clean.shape)
df_clean.head()
```

```python
monthly_rev = df_clean.groupby('YearMonth')['Revenue'].sum()
mom_growth = monthly_rev.pct_change() * 100
mom_growth.plot(kind='bar', figsize=(10,4), title='MoM Revenue Growth %')
```
